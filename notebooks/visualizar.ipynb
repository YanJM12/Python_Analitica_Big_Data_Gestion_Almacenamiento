```python

#Cargar la segunda hoja del diccionario
diccionario_datos_df = pd.read_excel( r"C:\Users\juan\Desktop\Trabajo final GA\data\bronze\Diccionario_de _datos _IPM_2025.xlsx",
    sheet_name=1,
    header=None
)

#Buscar la fila que contiene los encabezados
for i, fila in diccionario_datos_df.iterrows():
    valores = fila.astype(str).str.strip().str.lower().tolist()

    if "campo" in valores and "nombre" in valores:
        encabezados = valores
        diccionario_datos_df.columns = (
            fila.astype(str).str.strip().tolist()
        )
        diccionario_datos_df = diccionario_datos_df.iloc[i + 1:].copy()
        break

#Limpiar nombres de columnas
diccionario_datos_df.columns = (
    diccionario_datos_df.columns.astype(str).str.strip()
)

#Conservar las cinco columnas del diccionario
diccionario_datos_df = diccionario_datos_df[
    ["Campo", "Nombre", "Descripción", "Valor", "var_txt"]
].copy()

#Organizar los registros
diccionario_datos_df = diccionario_datos_df.dropna(
    subset=["Nombre"]
).reset_index(drop=True)

#Convertir el campo a número
diccionario_datos_df["Campo"] = pd.to_numeric(
    diccionario_datos_df["Campo"], errors="coerce"
)

display(diccionario_datos_df.head(10))
```

```python

#Limpiar nombres de las columnas
diccionario_datos_df.columns = (
    diccionario_datos_df.columns.astype(str).str.strip()
)

#Limpiar espacios en las columnas de texto
for columna in ["Nombre", "Descripción", "Valor", "var_txt"]:
    diccionario_datos_df[columna] = (
        diccionario_datos_df[columna]
        .astype("string")
        .str.strip()
    )

#Eliminar filas sin nombre de variable
diccionario_datos_df = diccionario_datos_df.dropna(
    subset=["Nombre"]
)

#Revisar registros duplicados
print(
    "Registros duplicados:",
    diccionario_datos_df.duplicated(subset=["Nombre"]).sum()
)

display(diccionario_datos_df.head())
```

```python

#Limpiar los nombres de las columnas del CSV
hogares_ipm_df.columns = (
    hogares_ipm_df.columns
    .str.replace('"', '', regex=False)
    .str.strip()
)

print(hogares_ipm_df.columns.tolist())
```

```python

# Revisar las variables registradas en el diccionario
display(
    diccionario_datos_df[
        ["Campo", "Nombre", "Descripción"]
    ].head(15)
)

# Comparar las columnas de la base con los campos del diccionario
columnas_base = set(hogares_ipm_df.columns.str.upper().str.strip())

campos_diccionario = set(
    diccionario_datos_df["Campo"]
    .dropna()
    .astype(str)
    .str.replace('"', "", regex=False)
    .str.strip()
    .str.upper()
)

coinciden = columnas_base.intersection(campos_diccionario)

print("Columnas de la base:", len(columnas_base))
print("Campos del diccionario:", len(campos_diccionario))
print("Variables que coinciden:", len(coinciden))
print("Ejemplos de coincidencias:", sorted(coinciden)[:20])
```

```python

# Crear una copia de la base original
hogares_ipm_descriptivo_df = hogares_ipm_df.copy()

# Normalizar los nombres para relacionarlos
hogares_ipm_descriptivo_df.columns = (
    hogares_ipm_descriptivo_df.columns.astype(str).str.strip().str.lower()
)

diccionario_datos_df["Nombre"] = (
    diccionario_datos_df["Nombre"].astype("string").str.strip().str.lower()
)

diccionario_datos_df["Descripción"] = (
    diccionario_datos_df["Descripción"].astype("string").str.strip()
)

# Crear el diccionario: nombre original -> descripción
renombrar_columnas = (
    diccionario_datos_df
    .dropna(subset=["Nombre", "Descripción"])
    .drop_duplicates(subset=["Nombre"])
    .set_index("Nombre")["Descripción"]
    .to_dict()
)

# Renombrar solo las columnas que tengan descripción disponible
hogares_ipm_descriptivo_df = hogares_ipm_descriptivo_df.rename(
    columns=renombrar_columnas
)

# Mostrar el resultado
print("Dimensiones de la base:", hogares_ipm_descriptivo_df.shape)
print("\nColumnas descriptivas:")
print(hogares_ipm_descriptivo_df.columns.tolist())

display(hogares_ipm_descriptivo_df.head())
```
